**[Task 2.2.]()** Implement the two functions `sobel_h` and `sobel_v` for horizontal and vertical $3 \times 3$ Sobel filtering, respectively:

```python
def sobel_h(img):
    img_gh = img.copy()
    for x in range(1, img.shape[0]-1) :
        for y in range(1, img.shape[1]-1) :
            img_gh[x, y] = 1/8 *(img[x-1, y-1] + 2*img[x-1, y] + img[x-1, y+1] - img[x+1, y-1] - 2*img[x+1, y] - img[x+1, y+1])
    return img_gh
    
def sobel_v(img):
    img_gh = img.copy()
    for x in range(1, img.shape[0]-1) :
            for y in range(1, img.shape[1]-1) :
                img_gh[x, y] = 1/8 *(img[x-1, y-1] + 2*img[x, y-1] + img[x+1, y-1] - img[x-1, y+1] - 2*img[x, y+1] - img[x+1, y+1])
    return img_gh
```

**[Task 2.3.]()** Test your implementations from above:

```python
cells_gh = sobel_h(cells)
plt.figure()
plt.imshow(cells_gh, cmap = "gray")
plt.colorbar()
```

```python
cells_gv = sobel_v(cells)
plt.figure()
plt.imshow(cells_gv, cmap = "gray")
plt.colorbar()
```

**[Task 2.4.]()** Compute the image gradient magnitude:

```python
cells_gm = np.sqrt(cells_gh**2 + cells_gv**2)
plt.figure()
plt.imshow(cells_gm, cmap ="gray")
plt.colorbar()
```
