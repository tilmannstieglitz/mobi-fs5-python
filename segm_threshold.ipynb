**[Task 1.2 (d).]()** Binarize the image by applying an intensity threshold:

```python
print(dna_0.shape)
```

```python
dna_0_bi = dna_0.copy()

for x in range(1024) :
    for y in range(1344) :
        if dna_0[x, y] > 0.10 :
            dna_0_bi[x, y] = 1
        else :
            dna_0_bi[x, y] = 0
```

```python
plt.figure()
plt.imshow(dna_0_bi, "gray")
plt.colorbar()
```

**[Task 1.2 (e).]()** Load the ground truth:

```python
ground_truth = plt.imread("data/NIH3T3/gt/0.png")
plt.figure()
plt.imshow(ground_truth, "gray")
plt.colorbar()
```

Compute and print the Dice coefficient:

```python
def dice_coefficient(picture, ground_truth):
    return (2 * (picture * ground_truth).sum()) / (picture.sum() + ground_truth.sum())
```

```python
dice_dna_0_gt = dice_coefficient(dna_0_bi, ground_truth)
dice_dna_0_gt
```
